```python
file_path = "data/stats.pdf"
loader = PyPDFLoader(file_path)
data = loader.load()
```

```python
question_gen = ""
for page in data:
    question_gen += page.page_content
```

```python
question_gen
```

```python
from langchain_text_splitters import TokenTextSplitter
```

```python
splitter_ques_gen = TokenTextSplitter(
    model_name="gpt-3.5-turbo",
    chunk_size=5000,
    chunk_overlap=200
)
```

```python
chunks_ques_gen = splitter_ques_gen.split_text(question_gen)
```

```python
chunks_ques_gen
```

```python
len(chunks_ques_gen)
```

```python
type(chunks_ques_gen)
```

```python
from langchain_community.docstore.document import Document
```

```python
document_ques_gen = [Document(page_content = t) for t in chunks_ques_gen]
```

```python
print(document_ques_gen)
```

```python
splitter_ans_gen=TokenTextSplitter(
    model_name="gpt-3.5-turbo",
    chunk_size=1000,
    chunk_overlap=100
)
```

```python
document_ans_gen = splitter_ans_gen.split_documents(document_ques_gen)
```

```python
document_ans_gen
```

```python
len(document_ans_gen)
```

```python
from langchain_groq import ChatGroq

llm_ques_gen_pipeline = ChatGroq(
    model="openai/gpt-oss-20b",
    temperature=0.3
)
```

```python
prompt_template = """
Generate 10 relevant, non-duplicate questions from the provided text. Cover its key information. Output questions only, numbered sequentially, one per line. No headings or answers.
Generate complete questions only. Never output an unfinished question. only 10 questions
Text:
{text}
"""
```

```python
from langchain_core.prompts import PromptTemplate
PROMPT_QUESTIONS = PromptTemplate(template=prompt_template, input_variables=['text'])
```

```python
refine_template = """
Refine the existing questions using the new text. Add important missing questions, remove duplicates, and keep all questions grounded in the source. Output the complete question list only, numbered sequentially, one per line. No headings or answers.
Generate complete questions only. Never output an unfinished question. only 10 questions
Existing questions:
{existing_answer}

New text:
{text}
"""
```

```python
REFINE_PROMPT_QUESTIONS = PromptTemplate(
    input_variables=['existing_answer', "text"],
    template=refine_template
)
```

```python
from langchain_classic.chains.summarize import load_summarize_chain
```

```python
ques_gen_chain = load_summarize_chain(llm = llm_ques_gen_pipeline,
                                        chain_type="refine",
                                        
                                        question_prompt=PROMPT_QUESTIONS,
                                        refine_prompt=REFINE_PROMPT_QUESTIONS
                                        )
```

```python
ques = ques_gen_chain.run(document_ques_gen)
print(ques)
```

```python
from langchain_huggingface import HuggingFaceEmbeddings
```

```python
embeddings = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)
```

```python
from langchain_community.vectorstores import FAISS
```

```python
vectorstore = FAISS.from_documents(
    document_ques_gen,
    embeddings
)
```

```python
llm_ans_gen_pipeline = ChatGroq(
    model="openai/gpt-oss-20b",
    temperature=0.3
)
```

```python
ques_list = ques.split("\n")
print(ques_list)
```

```python
from langchain_classic.chains import RetrievalQA
```

```python
answer_generation_chain = RetrievalQA.from_chain_type(
    llm=llm_ans_gen_pipeline,
    chain_type="stuff",
    retriever=vectorstore.as_retriever(
        search_kwargs={"k": 4}
    )
)
```

```python
for question in ques_list:
    print("Question:", question)
    answer = answer_generation_chain.run(question)
    print("Answer:", answer)
    print("-" * 50)

    with open("answer.txt", "a", encoding="utf-8") as f:
        f.write(f"Question: {question}\n")
        f.write(f"Answer: {answer}\n")
        f.write("-" * 70 + "\n\n")
```
